# Data Preprocessing

This notebook applies the complete preprocessing pipeline to the raw survey data and saves the cleaned dataset.

```python
import pandas as pd
import numpy as np
import sys
import os

# Add src directory to path
sys.path.append(os.path.abspath('../src'))

from preprocessing import full_preprocessing_pipeline

# Load the raw dataset
data_path = '../data/survey.csv'
df_raw = pd.read_csv(data_path)

print(f"Raw dataset loaded successfully!")
print(f"Shape: {df_raw.shape}")
```

## Apply Full Preprocessing Pipeline

```python
# Apply the complete preprocessing pipeline
df_cleaned = full_preprocessing_pipeline(df_raw)

print(f"\nPreprocessing complete!")
print(f"Final shape: {df_cleaned.shape}")
```

## Verify Preprocessing Results

```python
# Check for missing values
print("Missing Values After Preprocessing:")
print(df_cleaned.isnull().sum().sum())
print(f"\nAll columns have no missing values: {df_cleaned.isnull().sum().sum() == 0}")
```

```python
# Display data types
print("\nData Types After Preprocessing:")
print(df_cleaned.dtypes.value_counts())
```

```python
# Display column names
print(f"\nTotal Columns: {len(df_cleaned.columns)}")
print("\nColumn Names:")
for i, col in enumerate(df_cleaned.columns, 1):
    print(f"{i}. {col}")
```

```python
# Check Gender distribution after cleaning
if 'Gender_Male' in df_cleaned.columns:
    print("\nGender Distribution (One-Hot Encoded):")
    gender_cols = [col for col in df_cleaned.columns if col.startswith('Gender_')]
    for col in gender_cols:
        count = df_cleaned[col].sum()
        print(f"{col}: {count}")
```

```python
# Check Age statistics after cleaning
if 'Age' in df_cleaned.columns:
    print("\nAge Statistics After Cleaning:")
    print(df_cleaned['Age'].describe())
```

```python
# Check work_interfere encoding
if 'work_interfere' in df_cleaned.columns:
    print("\nWork Interfere Encoding:")
    print(df_cleaned['work_interfere'].value_counts().sort_index())
```

```python
# Check no_employees encoding
if 'no_employees' in df_cleaned.columns:
    print("\nNumber of Employees Encoding:")
    print(df_cleaned['no_employees'].value_counts().sort_index())
```

```python
# Check leave encoding
if 'leave' in df_cleaned.columns:
    print("\nLeave Encoding:")
    print(df_cleaned['leave'].value_counts().sort_index())
```

## Save Cleaned Dataset

```python
# Ensure data directory exists
os.makedirs('../data', exist_ok=True)

# Save cleaned dataset
output_path = '../data/cleaned_survey.csv'
df_cleaned.to_csv(output_path, index=False)

print(f"Cleaned dataset saved to: {output_path}")
print(f"Final shape: {df_cleaned.shape}")
```
